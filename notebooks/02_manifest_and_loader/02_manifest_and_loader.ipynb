# Road PCI Intelligence — Step 2: Manifests, Mask Conversion, Leak-Proof Splits & Metric Reconciliation

This notebook establishes the foundational data ingestion pipeline for the **Road PCI Intelligence** project:
1. **Manifest Construction**: Builds leak-proof manifests in `manifests/` from the real files on disk.
2. **Standardized Mask Conversion Rules**:
   - **`road-crack-dataset`**: Value 1 $\to$ Crack; Values 2 & 3 $\to$ Background (Construction joints and core drilling).
   - **`CRACK500`**: Any non-zero $\to$ Crack.
   - **`pothole`**: YOLOv8 polygon vector vertices rasterized to binary mask via `cv2.fillPoly`.
   - **`RDD2022ES`**: Bounding boxes only, reserved for downstream detection.
3. **Leak-Proof Group Partitioning**:
   - **Pothole**: Base group ID `pic-<N>` (augmented variations co-located).
   - **Road-Crack**: Group ID = scene (`road_<N>` / `Production_1_ortho_merge`). Scene-level 70/15/15 partition.
   - **CRACK500**: Preserves official benchmark split lists (`train.lst`, `val.lst`, `test.lst`).
4. **Verification Test Suite**: Validates zero group leakage, disk path existence, split counts, and reports RDD tests as SKIPPED when run in segmentation reconciliation mode.
5. **Foreground Pixel Metric Reconciliation**: Recomputes exact pixel-weighted and per-image foreground stats across actual masks, reconciles road-crack values {1, 2, 3}, and audits pothole polygon coverage with timing logs.

```python
# ==============================================================================
# 4. STAGE 3: FOREGROUND PIXEL METRIC RECONCILIATION & MASK AUDIT (ACTUAL MASKS)
# ==============================================================================
import time
import numpy as np
import pandas as pd
from PIL import Image
from collections import Counter
from src.data_rules import load_standardized_mask, rasterize_pothole_polygons
from src.build_manifest import DATASET_SPECS, resolve_dataset_root
import cv2

t0_stage3 = time.time()
seg_df = pd.read_csv('manifests/segmentation_manifest.csv')

print('=' * 100)
print('RECONCILIATION PART 1: PER-SOURCE & PER-SPLIT FOREGROUND FRACTIONS (ACTUAL MASKS)')
print('=' * 100)

split_stats_rows = []
sources = ['road_crack', 'crack500', 'pothole']
splits = ['train', 'val', 'test']

source_all_per_image_fractions = {}

for src_name in sources:
    spec = DATASET_SPECS[src_name]
    root = resolve_dataset_root(spec['owner'], spec['name'])
    if root is None:
        print(f'Warning: Dataset root for {src_name} not found.')
        continue
        
    source_all_per_image_fractions[src_name] = []
    
    for sp in splits:
        sub = seg_df[(seg_df['source'] == src_name) & (seg_df['split'] == sp)]
        n_images = len(sub)
        n_groups = sub['group_id'].nunique()
        
        split_total_pixels = 0
        split_fg_pixels = 0
        per_img_fractions = []
        
        for _, row in sub.iterrows():
            lbl_rel = row['mask_path_or_label_path']
            fg_img_cnt = 0
            img_pixels = int(row['width']) * int(row['height'])
            
            if pd.notna(lbl_rel) and lbl_rel != '':
                lbl_p = root / lbl_rel
                if lbl_p.exists():
                    mask = load_standardized_mask(src_name, lbl_p, img_shape=(int(row['height']), int(row['width'])))
                    h_act, w_act = mask.shape[:2]
                    img_pixels = h_act * w_act
                    fg_img_cnt = int(np.sum(mask == 1))
            
            split_total_pixels += img_pixels
            split_fg_pixels += fg_img_cnt
            img_pct = (fg_img_cnt / img_pixels) * 100.0 if img_pixels > 0 else 0.0
            per_img_fractions.append(img_pct)
            source_all_per_image_fractions[src_name].append(img_pct)
            
        pixel_weighted_pct = (split_fg_pixels / split_total_pixels * 100.0) if split_total_pixels > 0 else 0.0
        mean_per_img_pct = float(np.mean(per_img_fractions)) if per_img_fractions else 0.0
        median_per_img_pct = float(np.median(per_img_fractions)) if per_img_fractions else 0.0
        min_per_img_pct = float(np.min(per_img_fractions)) if per_img_fractions else 0.0
        max_per_img_pct = float(np.max(per_img_fractions)) if per_img_fractions else 0.0
        defect_type = 'Crack' if sub['labels_crack'].iloc[0] else 'Pothole'
        
        split_stats_rows.append({
            'Source': src_name,
            'Split': sp,
            'Images': n_images,
            'Groups': n_groups,
            'Defect': defect_type,
            'Total Pixels': split_total_pixels,
            'FG Pixels': split_fg_pixels,
            'Pixel-Weighted FG (%)': f'{pixel_weighted_pct:.3f}%',
            'Mean Per-Image FG (%)': f'{mean_per_img_pct:.3f}%',
            'Median Per-Image (%)': f'{median_per_img_pct:.3f}%',
            'Min Per-Image (%)': f'{min_per_img_pct:.3f}%',
            'Max Per-Image (%)': f'{max_per_img_pct:.3f}%'
        })

df_split_stats = pd.DataFrame(split_stats_rows)
print(df_split_stats.to_string(index=False))

print('\n' + '=' * 100)
print('RECONCILIATION PART 1 (CONT): OVERALL PER-SOURCE SUMMARY (ALL SPLITS COMBINED)')
print('=' * 100)
for src_name in sources:
    all_fracs = source_all_per_image_fractions[src_name]
    sub_src = seg_df[seg_df['source'] == src_name]
    tot_pix = sum([r['Total Pixels'] for r in split_stats_rows if r['Source'] == src_name])
    tot_fg = sum([r['FG Pixels'] for r in split_stats_rows if r['Source'] == src_name])
    pw_pct = (tot_fg / tot_pix * 100.0) if tot_pix > 0 else 0.0
    print(f'Source: {src_name} (Total Images: {len(all_fracs)}):')
    print(f'  • Total Pixels:          {tot_pix:,}')
    print(f'  • Foreground Pixels:     {tot_fg:,}')
    print(f'  • Pixel-Weighted FG (%): {pw_pct:.4f}%')
    print(f'  • Mean Per-Image FG (%): {np.mean(all_fracs):.4f}%')
    print(f'  • Median Per-Image (%):  {np.median(all_fracs):.4f}%')
    print(f'  • Min Per-Image (%):     {np.min(all_fracs):.4f}%')
    print(f'  • Max Per-Image (%):     {np.max(all_fracs):.4f}%')

# ==============================================================================
# RECONCILIATION PART 2: ROAD_CRACK RAW MASK VALUE COUNTS (ALL 435 MASKS)
# ==============================================================================
print('\n' + '=' * 100)
print('RECONCILIATION PART 2: ROAD_CRACK MULTI-CLASS MASK VALUE AUDIT OVER ALL 435 MASKS')
print('=' * 100)
rc_spec = DATASET_SPECS['road_crack']
rc_root = resolve_dataset_root(rc_spec['owner'], rc_spec['name'])
rc_sub = seg_df[seg_df['source'] == 'road_crack']

rc_val_counts = Counter()
rc_total_mask_pixels = 0

for _, row in rc_sub.iterrows():
    mask_p = rc_root / row['mask_path_or_label_path']
    m_arr = np.array(Image.open(mask_p))
    rc_total_mask_pixels += m_arr.size
    unq, cnts = np.unique(m_arr, return_counts=True)
    for u, c in zip(unq, cnts):
        rc_val_counts[int(u)] += int(c)

print(f'Total masks inspected: {len(rc_sub)}')
print(f'Total pixels across all 435 masks: {rc_total_mask_pixels:,}')
print('Pixel counts per raw mask value:')
for v in sorted(rc_val_counts.keys()):
    v_cnt = rc_val_counts[v]
    v_pct = (v_cnt / rc_total_mask_pixels) * 100.0
    print(f'  Value {v}: {v_cnt:,} pixels ({v_pct:.4f}%)')

val1_count = rc_val_counts[1]
val2_count = rc_val_counts[2]
val3_count = rc_val_counts[3]
earlier_audit_val1 = 3126349
match_val1 = (val1_count == earlier_audit_val1)
print(f'\nComparison with Earlier Audit (Value 1 = 3,126,349):')
print(f'  • Current Recomputed Value 1 Pixels: {val1_count:,}')
print(f'  • Earlier Audit Value 1 Pixels:      {earlier_audit_val1:,}')
print(f'  • Difference:                         {val1_count - earlier_audit_val1}')
print(f'  • Exactly Matches Earlier Audit:     {match_val1}')
tot_fg_road_crack = sum([r['FG Pixels'] for r in split_stats_rows if r['Source'] == 'road_crack'])
print(f'  • Manifest Converted Crack Pixels:   {tot_fg_road_crack:,} (Matches Value 1: {tot_fg_road_crack == val1_count})')

# ==============================================================================
# RECONCILIATION PART 3: POTHOLE COVERAGE STATS (ALL 780 IMAGES)
# ==============================================================================
print('\n' + '=' * 100)
print('RECONCILIATION PART 3: POTHOLE POLYGON COVERAGE AUDIT OVER ALL 780 IMAGES')
print('=' * 100)
pothole_spec = DATASET_SPECS['pothole']
p_root = resolve_dataset_root(pothole_spec['owner'], pothole_spec['name'])
p_sub = seg_df[seg_df['source'] == 'pothole']

cov_data_rules_round = []
cov_trunc_int = []
pothole_counts = []

for _, row in p_sub.iterrows():
    w, h = int(row['width']), int(row['height'])
    lbl_p = p_root / row['mask_path_or_label_path']
    
    # 1. Using data_rules.py (int(round(...)))
    mask_round = rasterize_pothole_polygons((h, w), lbl_p)
    cov_round = (np.sum(mask_round == 1) / (w * h)) * 100.0
    cov_data_rules_round.append(cov_round)
    
    # 2. Using truncation int(...)
    mask_trunc = np.zeros((h, w), dtype=np.uint8)
    n_holes = 0
    if lbl_p.exists() and lbl_p.stat().st_size > 0:
        with open(lbl_p, 'r', encoding='utf-8') as f:
            for line in f:
                parts = line.strip().split()
                if len(parts) > 4:
                    coords = [float(x) for x in parts[1:]]
                    pts = []
                    for i in range(0, len(coords), 2):
                        px = int(coords[i] * w)
                        py = int(coords[i + 1] * h)
                        pts.append([px, py])
                    if len(pts) >= 3:
                        pts_arr = np.array(pts, dtype=np.int32).reshape((-1, 1, 2))
                        cv2.fillPoly(mask_trunc, [pts_arr], color=1)
                        n_holes += 1
    cov_trunc = (np.sum(mask_trunc == 1) / (w * h)) * 100.0
    cov_trunc_int.append(cov_trunc)
    pothole_counts.append(n_holes)

print(f'Total Pothole Images Evaluated: {len(p_sub)}')
print(f'Mean Potholes Per Image:        {np.mean(pothole_counts):.2f}')
print(f'Median Potholes Per Image:      {np.median(pothole_counts):.1f}')
print('\nCoverage Statistics with data_rules.py (int(round(coord * dim))):')
print(f'  • Mean Per-Image Coverage:   {np.mean(cov_data_rules_round):.2f}%')
print(f'  • Median Per-Image Coverage: {np.median(cov_data_rules_round):.2f}%')
print(f'  • Min Per-Image Coverage:    {np.min(cov_data_rules_round):.2f}%')
print(f'  • Max Per-Image Coverage:    {np.max(cov_data_rules_round):.2f}%')

print('\nCoverage Statistics with int(coord * dim) truncation (as in 01_data_exploration):')
print(f'  • Mean Per-Image Coverage:   {np.mean(cov_trunc_int):.2f}%')
print(f'  • Median Per-Image Coverage: {np.median(cov_trunc_int):.2f}%')
print(f'  • Min Per-Image Coverage:    {np.min(cov_trunc_int):.2f}%')
print(f'  • Max Per-Image Coverage:    {np.max(cov_trunc_int):.2f}%')

earlier_pothole_mean = 15.98
earlier_pothole_med = 12.78
print(f'\nComparison with Earlier Audit (Mean = {earlier_pothole_mean}%, Median = {earlier_pothole_med}%):')
print(f'  • Current Mean:   {np.mean(cov_data_rules_round):.2f}% (Earlier: {earlier_pothole_mean}%) -> Delta: {np.mean(cov_data_rules_round) - earlier_pothole_mean:+.2f}%')
print(f'  • Current Median: {np.median(cov_data_rules_round):.2f}% (Earlier: {earlier_pothole_med}%) -> Delta: {np.median(cov_data_rules_round) - earlier_pothole_med:+.2f}%')
print('Explanation of metrics:')
print('  - The earlier audit reported the unweighted mean and median of per-image coverage percentages.')
print('  - The pixel-weighted fraction is: total_fg_pixels / total_pixels.')

t_stage3 = time.time() - t0_stage3
print(f'\n[STAGE 3/4] Mask scan & foreground reconciliation completed in {t_stage3:.2f}s')
```

```python
# ==============================================================================
# 5. STAGE 4: VISUAL INSPECTION SAMPLES
# ==============================================================================
import time
import random
import matplotlib.pyplot as plt
from PIL import Image

t0_stage4 = time.time()
print('=' * 80)
print('STAGE 4: VISUAL INSPECTION SAMPLES')
print('=' * 80)

random.seed(42)

COLOR_CRACK_RGB = np.array([255, 30, 30], dtype=np.uint8)    # Red
COLOR_POTHOLE_RGB = np.array([30, 120, 255], dtype=np.uint8)  # Blue

def colorize_binary_mask(mask: np.ndarray, color_rgb: np.ndarray) -> np.ndarray:
    h, w = mask.shape[:2]
    rgb = np.zeros((h, w, 3), dtype=np.uint8)
    rgb[mask == 1] = color_rgb
    return rgb

for src_name in ['road_crack', 'crack500', 'pothole']:
    spec = DATASET_SPECS[src_name]
    root = resolve_dataset_root(spec['owner'], spec['name'])
    if root is None:
        continue
        
    sub = seg_df[seg_df['source'] == src_name]
    sub_with_mask = sub[sub['mask_path_or_label_path'].notna() & (sub['mask_path_or_label_path'] != '')]
    sample_df = sub_with_mask.sample(min(4, len(sub_with_mask)), random_state=42)
    
    is_crack = sub['labels_crack'].iloc[0]
    color = COLOR_CRACK_RGB if is_crack else COLOR_POTHOLE_RGB
    defect_name = 'Crack' if is_crack else 'Pothole'
    color_name = 'Red' if is_crack else 'Blue'
    
    fig, axes = plt.subplots(4, 2, figsize=(11, 14))
    fig.suptitle(f'Dataset: {src_name} — 4 Visual Verification Samples\nDefect: {defect_name} (Color: {color_name})',
                 fontsize=14, weight='bold', y=0.995)
    
    for idx, (_, row) in enumerate(sample_df.iterrows()):
        img_p = root / row['image_path']
        lbl_p = root / row['mask_path_or_label_path']
        
        img = Image.open(img_p).convert('RGB')
        w, h = img.size
        mask = load_standardized_mask(src_name, lbl_p, img_shape=(h, w))
        colored_mask = colorize_binary_mask(mask, color)
        fg_pct = (np.sum(mask == 1) / mask.size) * 100.0
        
        stem = Path(row['image_path']).stem
        if len(stem) > 28:
            stem = stem[:25] + '...'
            
        # Left: Original Image
        ax_img = axes[idx, 0]
        ax_img.imshow(img)
        ax_img.set_title(f"[{row['split'].upper()}] Image: {stem}\n(Group: {row['group_id']})",
                         fontsize=9.5, pad=6)
        ax_img.axis('off')
        
        # Right: Converted Mask
        ax_mask = axes[idx, 1]
        ax_mask.imshow(colored_mask)
        ax_mask.set_title(f"[{row['split'].upper()}] Converted {defect_name} ({color_name})\nFG Area: {fg_pct:.2f}% | Size: {w}x{h}",
                          fontsize=9.5, pad=6)
        ax_mask.axis('off')
        
    plt.subplots_adjust(top=0.94, hspace=0.35, wspace=0.15)
    plt.show()

t_stage4 = time.time() - t0_stage4
print(f'\n[STAGE 4/4] Sample visual verification completed in {t_stage4:.2f}s')

total_time = t_stage1 + t_stage2 + t_stage3 + t_stage4
print(f'\n[TOTAL TIME] Notebook execution completed in {total_time:.2f}s')
```
